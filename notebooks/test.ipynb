```python
import pandas as pd
import plotly.express as px

data = pd.read_csv('../scores.csv')

data = data [["timestamp", "score"]]
print(data.columns)

# Create a line chart with Plotly
fig = px.line(
    data,
    x="timestamp",
    y="score",    
    labels={"timestamp": "Timestamp", "score": "Scores"}
)
fig.show()
```
